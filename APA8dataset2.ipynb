```python
# Convert important columns to numeric where appropriate
d2["year"] = pd.to_numeric(d2["year"], errors="coerce")
d2["district_code"] = pd.to_numeric(
    d2["district_code"],
    errors="coerce"
)
d2["state_code"] = pd.to_numeric(
    d2["state_code"],
    errors="coerce"
)
d2[target_col] = pd.to_numeric(
    d2[target_col],
    errors="coerce"
)

# Replace infinite values with missing values
d2 = d2.replace([np.inf, -np.inf], np.nan)

# Remove rows without essential information
d2 = d2.dropna(
    subset=[
        "year",
        "state",
        "district",
        target_col
    ]
).copy()

# Keep only positive rice-yield observations
d2 = d2[d2[target_col] > 0].copy()

# Convert year to integer
d2["year"] = d2["year"].astype(int)

# Remove exact duplicates
d2 = d2.drop_duplicates().reset_index(drop=True)

print("Cleaned Dataset 2 shape:", d2.shape)
display(d2.head())
```

```python
#Check the chronological split requirement
years = sorted(d2["year"].unique())

print("All available years:")
print(years)

print("\nNumber of unique years:", len(years))

if len(years) < 7:
    raise ValueError(
        "Dataset 2 does not contain the minimum seven unique years."
    )

test_years = years[-2:]
validation_years = years[-4:-2]
training_years = years[:-4]

print("\nTraining years:", training_years)
print("Validation years:", validation_years)
print("Test years:", test_years)
```

```python
#Create train, validation, and test datasets
train_d2 = d2[d2["year"].isin(training_years)].copy()
validation_d2 = d2[d2["year"].isin(validation_years)].copy()
test_d2 = d2[d2["year"].isin(test_years)].copy()

print("Training shape:", train_d2.shape)
print("Validation shape:", validation_d2.shape)
print("Test shape:", test_d2.shape)

print("\nTraining year range:")
print(train_d2["year"].min(), "to", train_d2["year"].max())

print("\nValidation years:")
print(sorted(validation_d2["year"].unique()))

print("\nTest years:")
print(sorted(test_d2["year"].unique()))
```

```python
#Verify leakage and split conditions
assert set(training_years).isdisjoint(validation_years)
assert set(training_years).isdisjoint(test_years)
assert set(validation_years).isdisjoint(test_years)

assert len(train_d2) > 0
assert len(validation_d2) > 0
assert len(test_d2) > 0

assert "rice_area_1000ha" not in []
assert "rice_production_1000tons" not in []

print("All chronological split checks passed.")
print("Train, validation, and test years are disjoint.")
```

```python
FEATURES_D2 = [
    "state",
    "district",
    "year"
]

TARGET_D2 = "rice_yield_kg_ha"

X_train_d2 = train_d2[FEATURES_D2].copy()
y_train_d2 = train_d2[TARGET_D2].copy()

X_validation_d2 = validation_d2[FEATURES_D2].copy()
y_validation_d2 = validation_d2[TARGET_D2].copy()

X_test_d2 = test_d2[FEATURES_D2].copy()
y_test_d2 = test_d2[TARGET_D2].copy()

print("Features:", FEATURES_D2)
print("Target:", TARGET_D2)

print("Training features:", X_train_d2.shape)
print("Validation features:", X_validation_d2.shape)
print("Test features:", X_test_d2.shape)
```

```python
#Import machine-learning libraries
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import joblib
import json
```

```python
#Define preprocessing
categorical_features_d2 = [
    "state",
    "district"
]

numeric_features_d2 = [
    "year"
]

categorical_pipeline_d2 = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

numeric_pipeline_d2 = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

preprocessor_d2 = ColumnTransformer([
    (
        "categorical",
        categorical_pipeline_d2,
        categorical_features_d2
    ),
    (
        "numeric",
        numeric_pipeline_d2,
        numeric_features_d2
    )
])

print("Preprocessing pipeline created successfully.")
```

```python
#Define the models
models_d2 = {
    "Median Baseline": Pipeline([
        ("preprocessor", preprocessor_d2),
        ("model", DummyRegressor(strategy="median"))
    ]),

    "Ridge Regression": Pipeline([
        ("preprocessor", preprocessor_d2),
        ("model", Ridge(alpha=1.0))
    ]),

    "Decision Tree": Pipeline([
        ("preprocessor", preprocessor_d2),
        (
            "model",
            DecisionTreeRegressor(
                max_depth=6,
                min_samples_leaf=10,
                random_state=42
            )
        )
    ]),

    "Random Forest": Pipeline([
        ("preprocessor", preprocessor_d2),
        (
            "model",
            RandomForestRegressor(
                n_estimators=60,
                max_depth=12,
                min_samples_leaf=5,
                n_jobs=2,
                random_state=42
            )
        )
    ])
}

print("Models created:")
for model_name in models_d2:
    print("-", model_name)
```

```python
#Train models and evaluate on validation data
validation_results_d2 = []

for model_name, model_pipeline in models_d2.items():
    print(f"Training: {model_name}")

    model_pipeline.fit(
        X_train_d2,
        y_train_d2
    )

    validation_predictions = model_pipeline.predict(
        X_validation_d2
    )

    mae = mean_absolute_error(
        y_validation_d2,
        validation_predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_validation_d2,
            validation_predictions
        )
    )

    r2 = r2_score(
        y_validation_d2,
        validation_predictions
    )

    validation_results_d2.append({
        "model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

validation_results_d2 = pd.DataFrame(validation_results_d2)

display(
    validation_results_d2.sort_values(
        "MAE",
        ascending=True
    )
)
```

```python
#Visualize validation model comparison
plt.figure(figsize=(10, 5))

sns.barplot(
    data=validation_results_d2.sort_values("MAE"),
    x="MAE",
    y="model"
)

plt.title("Validation MAE Comparison — Dataset 2")
plt.xlabel("Mean Absolute Error (Kg per ha)")
plt.ylabel("Model")
plt.tight_layout()
plt.show()
```

```python
#Select the best model
best_model_name_d2 = (
    validation_results_d2
    .sort_values("MAE", ascending=True)
    .iloc[0]["model"]
)

best_model_d2 = models_d2[best_model_name_d2]

print("Selected model:", best_model_name_d2)

best_validation_row_d2 = validation_results_d2[
    validation_results_d2["model"] == best_model_name_d2
]

display(best_validation_row_d2)
```

```python
#Evaluate the selected model on the test set
test_predictions_d2 = best_model_d2.predict(X_test_d2)

test_mae_d2 = mean_absolute_error(
    y_test_d2,
    test_predictions_d2
)

test_rmse_d2 = np.sqrt(
    mean_squared_error(
        y_test_d2,
        test_predictions_d2
    )
)

test_r2_d2 = r2_score(
    y_test_d2,
    test_predictions_d2
)

final_test_results_d2 = pd.DataFrame([{
    "model": best_model_name_d2,
    "MAE": test_mae_d2,
    "RMSE": test_rmse_d2,
    "R2": test_r2_d2
}])

display(final_test_results_d2)
```

```python
#Actual versus predicted values
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test_d2,
    test_predictions_d2,
    alpha=0.4
)

minimum_value = min(
    y_test_d2.min(),
    test_predictions_d2.min()
)

maximum_value = max(
    y_test_d2.max(),
    test_predictions_d2.max()
)

plt.plot(
    [minimum_value, maximum_value],
    [minimum_value, maximum_value],
    linestyle="--"
)

plt.title(
    f"Actual vs Predicted Rice Yield — {best_model_name_d2}"
)

plt.xlabel("Actual Rice Yield (Kg per ha)")
plt.ylabel("Predicted Rice Yield (Kg per ha)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#Test residual analysis
test_error_d2 = y_test_d2 - test_predictions_d2

plt.figure(figsize=(10, 5))

sns.histplot(
    test_error_d2,
    bins=40,
    kde=True
)

plt.axvline(
    0,
    linestyle="--"
)

plt.title("Test Residual Distribution — Dataset 2")
plt.xlabel("Residual: Actual - Predicted")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
#Error analysis table
error_analysis_d2 = test_d2[
    ["state", "district", "year", "rice_yield_kg_ha"]
].copy()

error_analysis_d2["predicted_yield_kg_ha"] = test_predictions_d2
error_analysis_d2["residual"] = (
    error_analysis_d2["rice_yield_kg_ha"]
    - error_analysis_d2["predicted_yield_kg_ha"]
)

error_analysis_d2["absolute_error"] = (
    error_analysis_d2["residual"].abs()
)

print("Largest prediction errors:")
display(
    error_analysis_d2
    .sort_values("absolute_error", ascending=False)
    .head(20)
)
```

```python
#Year-wise test performance
yearwise_results_d2 = []

for test_year in sorted(test_d2["year"].unique()):
    year_mask = test_d2["year"] == test_year

    actual_year = y_test_d2[year_mask]
    predicted_year = test_predictions_d2[year_mask]

    yearwise_results_d2.append({
        "year": int(test_year),
        "MAE": mean_absolute_error(
            actual_year,
            predicted_year
        ),
        "RMSE": np.sqrt(
            mean_squared_error(
                actual_year,
                predicted_year
            )
        ),
        "R2": r2_score(
            actual_year,
            predicted_year
        ) if len(actual_year) > 1 else np.nan,
        "sample_count": int(year_mask.sum())
    })

yearwise_results_d2 = pd.DataFrame(yearwise_results_d2)

display(yearwise_results_d2)
```

```python
#Plot year-wise MAE
plt.figure(figsize=(8, 5))

sns.barplot(
    data=yearwise_results_d2,
    x="year",
    y="MAE"
)

plt.title("Year-wise Test MAE — Dataset 2")
plt.xlabel("Year")
plt.ylabel("MAE (Kg per ha)")
plt.tight_layout()
plt.show()
```

```python
#Rolling-origin validation
rolling_origin_results_d2 = []

all_years_d2 = sorted(d2["year"].unique())

# Use three rolling validation windows
minimum_required_years = 7

if len(all_years_d2) >= minimum_required_years:
    rolling_windows_d2 = [
        {
            "train_end": all_years_d2[-6],
            "validation_year": all_years_d2[-5]
        },
        {
            "train_end": all_years_d2[-5],
            "validation_year": all_years_d2[-4]
        },
        {
            "train_end": all_years_d2[-4],
            "validation_year": all_years_d2[-3]
        }
    ]

    for window in rolling_windows_d2:
        rolling_train = d2[
            d2["year"] <= window["train_end"]
        ].copy()

        rolling_validation = d2[
            d2["year"] == window["validation_year"]
        ].copy()

        if len(rolling_train) == 0 or len(rolling_validation) == 0:
            continue

        rolling_model = Pipeline([
            ("preprocessor", ColumnTransformer([
                (
                    "categorical",
                    Pipeline([
                        (
                            "imputer",
                            SimpleImputer(
                                strategy="most_frequent"
                            )
                        ),
                        (
                            "encoder",
                            OneHotEncoder(
                                handle_unknown="ignore",
                                sparse_output=False
                            )
                        )
                    ]),
                    categorical_features_d2
                ),
                (
                    "numeric",
                    Pipeline([
                        (
                            "imputer",
                            SimpleImputer(strategy="median")
                        ),
                        (
                            "scaler",
                            StandardScaler()
                        )
                    ]),
                    numeric_features_d2
                )
            ])),
            ("model", best_model_d2.named_steps["model"])
        ])

        rolling_model.fit(
            rolling_train[FEATURES_D2],
            rolling_train[TARGET_D2]
        )

        rolling_predictions = rolling_model.predict(
            rolling_validation[FEATURES_D2]
        )

        rolling_mae = mean_absolute_error(
            rolling_validation[TARGET_D2],
            rolling_predictions
        )

        rolling_origin_results_d2.append({
            "train_end_year": int(window["train_end"]),
            "validation_year": int(window["validation_year"]),
            "train_samples": int(len(rolling_train)),
            "validation_samples": int(len(rolling_validation)),
            "MAE": float(rolling_mae)
        })

rolling_origin_results_d2 = pd.DataFrame(
    rolling_origin_results_d2
)

display(rolling_origin_results_d2)
```

```python
#Plot rolling-origin MAE
plt.figure(figsize=(9, 5))

sns.barplot(
    data=rolling_origin_results_d2,
    x="validation_year",
    y="MAE"
)

plt.title("Rolling-Origin Validation MAE — Dataset 2")
plt.xlabel("Validation Year")
plt.ylabel("MAE (Kg per ha)")
plt.tight_layout()
plt.show()
```
